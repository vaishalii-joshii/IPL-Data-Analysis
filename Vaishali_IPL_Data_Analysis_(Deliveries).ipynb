# INSIGHTS

I analysed the ball-by-ball IPL deliveries data to understand its quality, how innings progress, and how often wickets and extras occur. The key findings are below.

### 📌 Data Overview & Quality

- **Dataset size:** The data has 144,545 ball-by-ball records and 21 columns, covering team, batsman, bowler, runs and dismissal details for each delivery.

- **Core columns are complete:** Team names, innings and over numbers have no missing values, so they can be used directly for analysis.

- **One blank record:** The last row (match 611) has empty values in most columns, such as batsman, bowler and runs. This is an incomplete record and should be removed before any batsman or bowler analysis.

- **Expected nulls:** `player_dismissed` and `dismissal_kind` have 7,145 values, and `fielder` has 5,149. These columns are filled only when a wicket falls, so the nulls are not errors.

- **Data type note:** Because of the blank row, `ball` and the run columns were read as decimals (float). They should be converted to integers after cleaning.

### 📌 Wickets & Dismissals

- **Wickets are rare:** Only 7,145 of about 144,500 valid deliveries (about 4.9%) ended in a dismissal. On average, one wicket falls roughly every 20 balls.

- **Fielder involvement:** A fielder is recorded in 5,149 of 7,145 dismissals (about 72%). In the remaining 28% (such as bowled or LBW) no fielder is involved, and the bowler is the main contributor.

### 📌 Over-wise Analysis

- **Deliveries fall as the innings progresses:** Over 1 has the most deliveries (7,776). The count falls gradually to 6,302 in over 19 and drops sharply to 5,438 in over 20, about 30% fewer than over 1.

- **Why this happens:** Many innings end before 20 overs, through all-outs or because the chasing team reaches the target early. Fewer matches therefore reach the final overs.

- **Consistent early overs:** Overs 1–10 each have between 7,400 and 7,800 deliveries. This shows the first half of the innings is stable and almost every match goes through it.

### 📌 Ball-wise Analysis & Extras

- **Normal balls are evenly spread:** Balls 1 to 6 each appear about 23,100–23,400 times, as expected in a regular over.

- **Extra deliveries:** Balls 7, 8 and 9 appear 4,924 times in total (about 3.4% of deliveries). These are re-bowled balls caused by wides and no-balls.

- **What this means for bowlers:** About 1 in 30 deliveries is an extra ball, which gives batting teams free runs and extra scoring chances. Bowling discipline is therefore an important factor in match results.

### 📌 Conclusion

The deliveries data is detailed and mostly clean, with one blank record that must be removed. Wickets are rare (about 5%), and deliveries decline in the later overs because many innings end early. Extras make up about 3.4% of balls. After cleaning, this dataset can support deeper analysis such as top batsmen, top bowlers, runs per over, and dismissal types.
